```python
from keras import Sequential
from keras.layers import Input, Dense,  SimpleRNN
```

```python
model = Sequential()

model.add(Input(shape=(4,5)))
model.add(SimpleRNN(3))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
print(model.get_weights()[0].shape)
model.get_weights()[0]
```
